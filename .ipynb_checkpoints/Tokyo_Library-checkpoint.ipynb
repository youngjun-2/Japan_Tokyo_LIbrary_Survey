```python
import pandas as pd

df = pd.read_excel('Library.xlsx', skiprows=3, index_col=0, usecols='A:B')
```

```python
df.columns = ['2023년 개관 수']

df.sort_values('2023년 개관 수', ascending=False, inplace=True)

df.index.name = '구역'

df.drop(index='全国', inplace=True)
```

```python
df
```

```python
import matplotlib.pyplot as plt

import matplotlib

matplotlib.rcParams['font.family'] = 'AppleGothic', 'Hiragino Sans' # 한국어 설정 코드

matplotlib.rcParams['font.size'] = 10 # 글자 크기 설정

matplotlib.rcParams['axes.unicode_minus'] = False # 한글의 - 부호가 깨지는 것을 방지
```

```python
plt.figure(figsize=(14,7), dpi=300)

plt.bar(df.index, df['2023년 개관 수'], label='도서관 개관 수')
plt.xticks(rotation=45)
plt.legend(title='2023년도')

plt.savefig('Library.png')
```
